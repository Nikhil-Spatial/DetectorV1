```python
annotations_file_trainval = Path("../data/preprocessed/trainval/annotations.csv")
img_dir_trainval = Path("../data/preprocessed/trainval/Images")

trainval_dataset = ImageDataset(annotations_file_trainval, img_dir_trainval)
```

```python
mean_c1 = mean_c2 = mean_c3 = 0
std_c1 = std_c2 = std_c3 = 0
```

```python
import torch 

for image, target in trainval_dataset:
    image = image.flatten(1, -1).to(torch.float32)

    mean_c1 += torch.mean(image[0])
    mean_c2 += torch.mean(image[1])
    mean_c3 += torch.mean(image[2])

mean_c1 /= len(trainval_dataset)
mean_c2 /= len(trainval_dataset)
mean_c3 /= len(trainval_dataset)
```

```python
mean_c1, mean_c2, mean_c3
```

```python
for image, target in trainval_dataset:
    image = image.flatten(1, -1).to(torch.float32)

    std_c1 += torch.sum((image[0] - mean_c1).square())
    std_c2 += torch.sum((image[1] - mean_c2).square())
    std_c3 += torch.sum((image[2] - mean_c3).square())
```

```python
std_c1, std_c2, std_c3
```

```python
pixels_in_one_channel = len(trainval_dataset) * 50176

std_c1 /= pixels_in_one_channel
std_c2 /= pixels_in_one_channel
std_c3 /= pixels_in_one_channel
```

```python
std_c1 = std_c1.sqrt() 
std_c2 = std_c2.sqrt() 
std_c3 = std_c3.sqrt() 

mean_c1 = mean_c1 
mean_c2 = mean_c2 
mean_c3 = mean_c3 
```
